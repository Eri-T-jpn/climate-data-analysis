本プログラムは、日本の主要6地点（札幌、仙台、東京、大阪、福岡、那覇）における気温偏差データの相関を分析し、「日本列島全体の気温変動に共通して見られる主要なパターン」を数理的に抽出することを目的としている。

```python
import numpy as np
```

```python
#データを読み込む
#6地点（札幌、仙台、東京、大阪、福岡、那覇）の気温偏差データが含まれるファイルをロード
ds = np.load('jp_tempa.npz')
#各地点の時系列データ（偏差）を抽出
sapporoa = ds['sapporoa']
sendaia = ds['sendaia']
tokyoa = ds['tokyoa']
osakaa = ds['osakaa']
fukuokaa = ds['fukuokaa']
nahaa = ds['nahaa']

# データの長さ（月数）を取得
N = tokyoa.shape[0]
#Nヶ月分の6地域の気温変化時系列を縦に並べた6行N列の行列
X = np.array([sapporoa,sendaia,tokyoa,osakaa,fukuokaa,nahaa])
# 行列Xとその転置行列の積をサンプル数Nで割り、共分散行列Cを求める
C = X@X.T/N

#共分散行列Cに対して固有値分解を行い、固有値Lamと固有ベクトルEを算出
[Lam,E] = np.linalg.eig(C)

#固有値が分散の大きさに対応するため、寄与率の高い順に並び替える
index = Lam.argsort()[::-1] #降順のインデックスを取得
Lam = Lam[index] #固有値を並び替え
E = E[:,index] #対応する固有ベクトル（主成分ベクトル）を並び替え

#固有値を対角成分に持つ行列Lを作成
L = np.diag(Lam)

#E：各列が主成分ベクトル（各地点の重み係数）
print(E)
#L：対角成分が各主成分の固有値（分散の大きさ）
print(L)
```
